```python
#THREE FUNCTION
def normalise(values, minimum=None, maximum=None):
    """
    Scale a list of numbers into the 0–1 range.

    Arguments:
        values: A list of numbers to normalise.
        minimum: Optional minimum value to use as the lower bound.
        maximum: Optional maximum value to use as the upper bound.

    Returns:
        A list of normalised numbers between 0 and 1.
        If all values are identical, returns a list of 0s.
    """

    # Use the supplied bounds, or calculate them from the values
    if minimum is None:
        minimum = min(values)

    if maximum is None:
        maximum = max(values)

    # Avoid division by zero when minimum and maximum are the same
    if maximum == minimum:
        return [0 for value in values]

    # Scale each value into the 0–1 range
    return [(value - minimum) / (maximum - minimum) for value in values]
```

```python
# Test with different values
values1 = [12, 18, 25, 40]
print(normalise(values1))

# Test when every value is identical
values2 = [5, 5, 5, 5]
print(normalise(values2))
```
