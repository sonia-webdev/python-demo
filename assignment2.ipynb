```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv("data/raw/penguins_size.csv")

# Display the first 5 rows
df.head()
```

```python
#dataset shape
print("dataset shape:")
print(df.shape)

#colimn names
print("\ncolumn names:")
print(df.columns.tolist())

#data types
print("\nData types:")
print(df.dtypes)

#null value per column
print("\nNull counts per column:")
print(df.isnull().sum())

#unique value per column
print("\nunique-value counts per column:")
print(df.nunique())

```

```python
#check for duplicate rows
print("number of duplicate rows:",df.duplicated().sum())
```

```python
#clean the missing values
#create a copy of orignl dataset for cleaning

clean_df=df.copy()

#fill missing numeric value with the median

numeric_columns=[
    "culmen_length_mm",
    "culmen_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]

for column in numeric_columns:
    clean_df[column] = clean_df[column].fillna(clean_df[column].median())

 #fill missing values in the sex column with the most commmon value
clean_df["sex"] = clean_df["sex"].fillna(clean_df["sex"].mode()[0])

#no duplicate rows were found,so no rows were removed
print("duplicate rows:",clean_df.duplicated().sum())
      
 #check missing values after cleaning

print("\nmissing values after cleaning:")
print(clean_df.isnull().sum())

#check data types
print("\ndata types after cleaning:")
print(clean_df.dtypes)
```

```python
#priny the shape before the groupby/merge transform
print("shape before transformm:",clean_df.shape)

#calculate aerage body mass and flipper length for each species
species_stats = clean_df.groupby("species").agg(averag_body_mass=("body_mass_g","mean"),
average_flipper_length=("flipper_length_mm","mean")).reset_index()

#dispay the grouped staistic
print("\nspecies statistics:")
print(species_stats)

#merge the species statistics back into the main dataframe
clean_df = clean_df.merge(species_stats, on="species",how="left")

#print the shape after transform
print("\nshape after transform:",clean_df.shape)

#dispay the first 5 rows
clean_df.head()

```

```python
#print shape before the pivot transform
print("shape of cleaned dataframe before pivot:",clean_df.shape)

#create a pivot table showing average body mass by species and sex
body_mass_pivot = clean_df.pivot_table(
    values="body_mass_g",
    index="species",
    columns="sex",
    aggfunc="mean"
)
print("\naverage body mass by species and sex:")
print(body_mass_pivot)

#print the shape of the pivot table
print("\nshape of pivot table:",body_mass_pivot.shape)
```
